```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


np.random.seed(42)

sem1 = np.random.uniform(6.0,9.5,60)
sem2 = sem1 + np.random.normal(0,0.35,60)
sem3 = sem2 + np.random.normal(0,0.30,60)
sem4 = sem3 + np.random.normal(0,0.25,60)
sem5 = sem4 + np.random.normal(0,0.20,60)

df = pd.DataFrame({

    "SEM 1":np.round(sem1,2),
    "SEM 2":np.round(sem2,2),
    "SEM 3":np.round(sem3,2),
    "SEM 4":np.round(sem4,2),
    "SEM 5":np.round(sem5,2)

})

print(df.head())


plt.figure(figsize=(8,6))

sns.heatmap(

    df.corr(),
    annot=True,
    cmap="coolwarm",
    linewidths=1,
    fmt=".2f"

)

plt.title("Correlation Heatmap",fontsize=13,fontweight="bold")

plt.tight_layout()

plt.show()


X = df[["SEM 2"]]

y = df["SEM 5"]

model = LinearRegression()

model.fit(X,y)

prediction = model.predict(X)



plt.figure(figsize=(8,4.5),dpi=100)

plt.scatter(

    X,
    y,

    color="#2b5c8f",

    edgecolor="white",

    linewidth=0.8,

    s=60,

    alpha=0.8,

    label="Actual Marks"

)

plt.plot(

    X,

    prediction,

    color="#e74c3c",

    linewidth=2.5,

    label="Regression Line"

)

plt.title(

    "Simple LR: SEM 2 vs SEM 5",

    fontsize=12,

    fontweight="bold"

)

plt.xlabel("SEM 2")

plt.ylabel("SEM 5")

plt.grid(True,linestyle=":",alpha=0.6)

plt.legend()

plt.tight_layout()

plt.show()



mae = mean_absolute_error(y,prediction)

mse = mean_squared_error(y,prediction)

rmse = np.sqrt(mse)

r2 = r2_score(y,prediction)

print("MAE :",round(mae,4))
print("MSE :",round(mse,4))
print("RMSE :",round(rmse,4))
print("R2 :",round(r2,4))



metrics=["MAE","MSE","RMSE","R2"]

values=[mae,mse,rmse,r2]

colors=["#00876c","#6aaa96","#aecdc2","#f19e7d"]

plt.figure(figsize=(8,4.5),dpi=100)

bars=plt.barh(

    metrics,

    values,

    color=colors,

    edgecolor="none",

    height=0.55

)

for bar in bars:

    w=bar.get_width()

    plt.text(

        w+0.01,

        bar.get_y()+bar.get_height()/2,

        f"{w:.4f}",

        va="center",

        fontsize=9,

        fontweight="bold"

    )

plt.title("Simple LR Performance Metrics",fontsize=12,fontweight="bold")

plt.grid(axis="x",linestyle=":",alpha=0.6)

plt.tight_layout()

plt.show()
```

```python
from sklearn.model_selection import train_test_split

X = df[["SEM 1", "SEM 2", "SEM 3", "SEM 4"]]
y = df["SEM 5"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

mlr = LinearRegression()
mlr.fit(X_train, y_train)

prediction = mlr.predict(X_test)



plt.figure(figsize=(8,4.5), dpi=100)

plt.scatter(
    y_test,
    prediction,
    color="#6c5ce7",
    edgecolor="white",
    linewidth=0.8,
    s=60,
    alpha=0.80,
    label="Predictions"
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    color="#00b894",
    linewidth=2,
    linestyle="--",
    label="Ideal Fit Line"
)

plt.title(
    "Multiple LR: Actual vs Predicted",
    fontsize=12,
    fontweight="bold",
    color="#2d3436"
)

plt.xlabel(
    "Actual Marks",
    fontsize=10,
    color="#636e72"
)

plt.ylabel(
    "Predicted Marks",
    fontsize=10,
    color="#636e72"
)

plt.grid(True, linestyle=":", alpha=0.6)

plt.legend(
    frameon=True,
    facecolor="#f8f9fa",
    edgecolor="none"
)

plt.tight_layout()
plt.show()


mae = mean_absolute_error(y_test, prediction)
mse = mean_squared_error(y_test, prediction)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, prediction)

print("\nMultiple Linear Regression Results")
print("----------------------------------")
print("MAE  :", round(mae,4))
print("MSE  :", round(mse,4))
print("RMSE :", round(rmse,4))
print("R2   :", round(r2,4))


metrics = ["MAE","MSE","RMSE","R2"]
values = [mae,mse,rmse,r2]

colors = [
    "#00876c",
    "#6aaa96",
    "#aecdc2",
    "#f19e7d"
]

plt.figure(figsize=(8,4.5), dpi=100)

bars = plt.barh(
    metrics,
    values,
    color=colors,
    edgecolor="none",
    height=0.55
)

plt.xlim(0, max(values)*1.15)

for bar in bars:

    width = bar.get_width()

    plt.text(
        width+0.01,
        bar.get_y()+bar.get_height()/2,
        f"{width:.4f}",
        fontsize=9,
        fontweight="bold",
        va="center",
        color="#2d3436"
    )

plt.title(
    "Multiple LR Performance Metrics",
    fontsize=12,
    fontweight="bold"
)

plt.xlabel("Metric Value")
plt.ylabel("Evaluation Metrics")

plt.grid(
    axis="x",
    linestyle=":",
    alpha=0.6
)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(12, 8), dpi=100)

semester_colors = [
    "#74b9ff",
    "#55efc4",
    "#ffeaa7",
    "#fab1a0",
    "#a29bfe"
]

semester_names = [
    "SEM 1",
    "SEM 2",
    "SEM 3",
    "SEM 4",
    "SEM 5"
]

positions = [
    (0,0),
    (0,1),
    (0,2),
    (1,0),
    (1,1)
]

for i in range(5):

    r, c = positions[i]

    sns.boxplot(
        y=df[semester_names[i]],
        ax=axes[r, c],
        color=semester_colors[i],
        width=0.45,
        linewidth=1.2
    )

    axes[r,c].set_title(
        semester_names[i],
        fontsize=11,
        fontweight="bold"
    )

    axes[r,c].set_ylabel("Marks")

    axes[r,c].grid(
        linestyle=":",
        alpha=0.5
    )

axes[1,2].axis("off")

plt.suptitle(
    "Semester Marks Distribution",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.show()


average_marks = [
    df["SEM 1"].mean(),
    df["SEM 2"].mean(),
    df["SEM 3"].mean(),
    df["SEM 4"].mean(),
    df["SEM 5"].mean()
]

plt.figure(figsize=(8,4.5), dpi=100)

bars = plt.bar(
    semester_names,
    average_marks,
    color=[
        "#74b9ff",
        "#55efc4",
        "#ffeaa7",
        "#fab1a0",
        "#a29bfe"
    ]
)

for bar in bars:

    h = bar.get_height()

    plt.text(
        bar.get_x()+bar.get_width()/2,
        h+0.03,
        f"{h:.2f}",
        ha="center",
        fontsize=9,
        fontweight="bold"
    )

plt.title(
    "Average Semester Marks",
    fontsize=12,
    fontweight="bold"
)

plt.xlabel("Semester")

plt.ylabel("Average Marks")

plt.grid(
    axis="y",
    linestyle=":",
    alpha=0.5
)

plt.tight_layout()

plt.show()

print("="*50)
print("ALL GRAPHS GENERATED SUCCESSFULLY")
print("="*50)
```
